```python
## importando o df
import sys
sys.path.append('..')

from src.limpeza import carregar_e_limpar

df = carregar_e_limpar()
```

```python
#importanto as libs
from sklearn.model_selection import train_test_split

#dividindo os dados
X = df[['age','pclass']]
y = df['fare']
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.33, random_state=42)

```

```python
#linar regretion
from sklearn.linear_model import LinearRegression
#treinando o modelo
modLinearReg = LinearRegression().fit(X_train,y_train)
```

```python
#testando o modelo
prev = modLinearReg.predict(X_test)
print(prev)
```

```python
#verificando taxa de acerto
from sklearn.metrics import mean_absolute_error, r2_score

result_mae = mean_absolute_error(y_test, prev)
result_r2 = r2_score(y_test, prev)

print(f"O resutado do MAE {result_mae} e o resultado do r2 {result_r2}")
```
