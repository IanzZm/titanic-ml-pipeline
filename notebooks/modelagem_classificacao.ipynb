```python
import sys
sys.path.append('..')

from src.limpeza import carregar_e_limpar

df = carregar_e_limpar()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score



df['sex'] = df['sex'].map({'male': 0, 'female': 1})


X, y = df[['age','pclass','fare','sex']], df['survived']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.33, random_state=42)
```

```python
#treinando regressao logistica
regression_log = LogisticRegression(random_state=0).fit(X_train,y_train)


prev = regression_log.predict(X_test)

#metricas
result_f1m = f1_score(y_test, prev, average='macro')
result_f1mi = f1_score(y_test, prev, average='micro')
result_f1w = f1_score(y_test, prev, average='weighted')

result_ac = accuracy_score(y_test, prev)
result_acNF = accuracy_score(y_test, prev, normalize=False)
```

```python
print(result_ac,result_acNF, result_f1m, result_f1mi, result_f1w)
```

```python
from sklearn.metrics import confusion_matrix

matrix_de_confusao = confusion_matrix(y_test, prev)

print(matrix_de_confusao)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# treinando arvore de decisao
dt = DecisionTreeClassifier(random_state=0)
dt.fit(X_train, y_train)
prev_dt = dt.predict(X_test)
rf1_dt = f1_score(y_test, prev_dt,  average='macro')
print(rf1_dt)

#treinando randomforest

drf = RandomForestClassifier(random_state=0).fit(X_train,y_train)
prev_drf = drf.predict(X_test)
rf1_drf = f1_score(y_test, prev_drf, average='macro')
print(rf1_drf)
```
