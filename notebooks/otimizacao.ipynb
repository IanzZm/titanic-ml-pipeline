```python
import sys
sys.path.append('..')

from src.limpeza import carregar_e_limpar

df = carregar_e_limpar()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score






X, y = df[['age','pclass','fare','sex']], df['survived']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.33, random_state=42)
```

```python
hiperparametros = {"n_estimators":[10, 50, 100, 150, 200], "max_depth": [None, 5, 10 , 15 , 20], "criterion":["gini","entropy","log_loss"]}
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_predict, GridSearchCV
model = RandomForestClassifier(random_state=42)
result = GridSearchCV(model, hiperparametros, cv=5)
result = result.fit(X_train,y_train)
```

```python
resultado = RandomForestClassifier(random_state=0).fit(X_train, y_train)
```

```python
prev_gs = result.predict(X_test)
prev = resultado.predict(X_test)

from sklearn.metrics import f1_score

f1_gs = f1_score(y_test,prev_gs, average="macro")
f1 = f1_score(y_test, prev, average="macro")

print(f1_gs)
print(f1)
```
